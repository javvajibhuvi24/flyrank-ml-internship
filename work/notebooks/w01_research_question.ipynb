## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import sys
import subprocess

REPO_URL = "https://github.com/javvajibhuvi24/flyrank-ml-internship"
REPO_DIR = "/content/flyrank-ml-internship"

if not os.path.isdir(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

import pandas as pd

csv_path = os.path.join(
    REPO_DIR,
    "data",
    "raw",
    "content_refresh_anonymized.csv"
)

df = pd.read_csv(csv_path)

df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

print("Number of pages:", len(df))
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
print(
    "Median word count - declining:",
    int(df.loc[df["trend_direction"] == "down", "word_count"].median())
)
print(
    "Median word count - growing:",
    int(df.loc[df["trend_direction"] == "up", "word_count"].median())
)
```

```python
#The starter dataset contains 30,000 pages, with 54.2% labeled as declining using the starter's current-window proxy. The median word count is 2,909 for declining pages and 2,848 for growing pages. The similar word counts suggest that content length alone may not explain the difference, making it useful to investigate multiple observable signals for prioritizing pages for review.
```
